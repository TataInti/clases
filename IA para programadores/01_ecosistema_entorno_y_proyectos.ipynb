## 1. Qué llamamos Inteligencia Artificial

En este curso usaremos una definición operativa:

> Un sistema de Inteligencia Artificial es software diseñado para
> realizar tareas que requieren interpretar datos, reconocer
> patrones, hacer predicciones, generar contenido o seleccionar
> acciones.

Esta definición no supone conciencia ni comprensión humana. Un
sistema puede producir una respuesta útil sin “entender” el problema
como lo haría una persona.

Todo sistema que veremos puede describirse con el mismo flujo:

```text
problema del negocio
        |
        v
datos de entrada
        |
        v
reglas o modelo
        |
        v
salida calculada
        |
        v
decisión o acción del software
```

Por ejemplo:

```text
imagen de una parcela -> detector -> cajas de malezas -> revisión
audio de una consulta  -> reconocimiento de voz -> texto -> registro
pregunta de un usuario -> búsqueda + modelo de lenguaje -> respuesta
```

La IA es una parte del sistema. La captura de datos, la interfaz,
las validaciones y la decisión final siguen siendo software
convencional.

### Software tradicional, reglas y aprendizaje

| Enfoque | Qué escribe el equipo | Qué cambia con los datos | Ejemplo |
|---|---|---|---|
| Software tradicional | Pasos exactos | Nada, salvo que cambiemos el código | Calcular un impuesto |
| Sistema de reglas | Condiciones explícitas | Las reglas deben editarse manualmente | Rechazar un campo inválido |
| Machine Learning | Datos, objetivo y proceso de entrenamiento | El modelo ajusta parámetros | Estimar riesgo |
| Modelo generativo | Instrucción, contexto y controles | Genera una secuencia probable | Resumir un documento |

No existe una jerarquía donde “más IA” signifique automáticamente
“mejor solución”. Si una regla exacta resuelve el problema de forma
estable, usar un modelo puede sumar costo y errores innecesarios.

## 2. Mapa del ecosistema de IA

```
Inteligencia Artificial
├── Sistemas de reglas y búsqueda
├── Machine Learning
│   ├── Algoritmos clásicos
│   │   ├── regresiones
│   │   ├── árboles
│   │   └── métodos por vecinos
│   └── Deep Learning
│       ├── redes para imágenes
│       ├── redes para audio
│       └── transformers
│           └── LLMs
└── Sistemas con agentes
    └── modelo + instrucciones + estado + herramientas
```

### Machine Learning

Machine Learning, o ML, es una forma de construir modelos a partir
de ejemplos. En lugar de escribir cada condición, definimos:

- qué información recibe el sistema;
- cuál es la salida esperada;
- qué ejemplos históricos puede usar;
- cómo mediremos sus errores.

Un algoritmo de entrenamiento busca parámetros que funcionen bien
sobre esos ejemplos.

### Deep Learning

Deep Learning es una familia dentro de ML basada en redes neuronales
con varias capas. Resulta especialmente útil cuando las entradas
contienen muchos valores y estructura compleja, como píxeles,
muestras de audio o tokens.

### Transformers y LLMs

Un transformer es una arquitectura de red que relaciona elementos
de una secuencia mediante attention. Un Large Language Model, o LLM,
es un modelo de lenguaje grande, habitualmente basado en
transformers, entrenado para estimar continuaciones de tokens.

```text
texto -> tokens -> transformer -> probabilidades -> próximo token
                                            |
                                            +-> repetir
```

El resultado puede parecer una explicación razonada, pero el
mecanismo básico sigue siendo una predicción estadística de
secuencias.

### Agentes

Un agente no es una nueva clase de modelo. Es un sistema que permite
que un modelo proponga pasos y solicite herramientas:

```text
objetivo -> modelo -> acción propuesta -> validación
                          |                  |
                          v                  v
                     herramienta --------> resultado
                          ^                  |
                          |__________________|
```

El software debe decidir qué herramientas existen, qué permisos
tienen y cuándo una acción requiere confirmación.

**IA no es sinónimo de LLM.** Tampoco todo problema necesita un
modelo: si una regla es exacta, estable y suficiente, suele ser la
solución más simple.

## 3. Algoritmo, modelo y datos: tres conceptos diferentes

Estos términos suelen mezclarse, pero cumplen funciones distintas:

| Concepto | Qué es | Ejemplo |
|---|---|---|
| Datos | Casos observados y sus variables | Temperatura y vibración históricas |
| Algoritmo | Procedimiento para ajustar parámetros | Árbol de decisión |
| Modelo | Resultado entrenado que recibe entradas y produce salidas | Árbol con preguntas ya aprendidas |
| Parámetro | Valor que el algoritmo aprende | Umbral de una rama |
| Hiperparámetro | Decisión configurada por el equipo | Profundidad máxima del árbol |

El flujo completo es:

```text
                  ENTRENAMIENTO
ejemplos + respuestas + algoritmo
                |
                v
        modelo con parámetros
                |
    guardar y versionar
                |
                v
                   INFERENCIA
entrada nueva -> modelo -> predicción
```

- Durante **entrenamiento**, el algoritmo puede comparar
  predicciones con respuestas conocidas y ajustar parámetros.
- Durante **inferencia**, el modelo ya entrenado recibe un caso nuevo.
  No debería modificar sus parámetros.
- Durante **evaluación**, reservamos ejemplos que el modelo no usó
  para aprender y medimos cómo generaliza.

Entrenamiento e inferencia pueden ocurrir en máquinas, momentos y
servicios diferentes.

## 4. La salida define el tipo de tarea

Antes de elegir una librería o una arquitectura debemos definir qué
salida necesita consumir el resto del software.

| Tarea | Entrada posible | Salida | Pregunta de ejemplo |
|---|---|---|---|
| Clasificación | Una fila o un texto | Una categoría | ¿Es urgente o no? |
| Regresión | Variables numéricas | Un número continuo | ¿Cuánta demanda habrá? |
| Detección | Una imagen | Clases, cajas y scores | ¿Qué objetos hay y dónde? |
| Segmentación | Una imagen | Una clase por píxel | ¿Qué superficie es cultivo? |
| Recuperación | Una consulta | Fragmentos ordenados | ¿Qué documento es relevante? |
| Generación | Un prompt y contexto | Una secuencia nueva | ¿Cómo resumimos este texto? |
| ASR: reconocimiento automático de voz | Una señal de audio | Texto | ¿Qué se dijo en la llamada? |
| TTS: síntesis de voz desde texto | Texto | Una señal de audio | ¿Cómo sintetizamos la respuesta? |

Una frase como “queremos usar IA con imágenes” todavía no define un
problema. Podría significar:

```text
una clase para toda la imagen       -> clasificación
varios objetos con su ubicación     -> detección
una etiqueta para cada píxel        -> segmentación
una descripción escrita             -> generación multimodal
```

La tarea determina los datos, el modelo, las métricas y la forma de
integrar la salida con una aplicación. También determina el tipo de
etiqueta que deberemos producir para entrenar y evaluar.

## 6. Ciclo de vida de un proyecto de IA

Un modelo aislado no es una solución completa. Un proyecto de IA
conecta una necesidad del negocio con datos, experimentos y
componentes de software que deben mantenerse.

```
problema
   |
   v
definición de la salida y del error aceptable
   |
   v
datos -> exploración -> preparación
   |                       |
   |                       v
   +-----------------> baseline
                           |
                           v
                    entrenamiento
                           |
                           v
                     evaluación
                           |
               ¿cumple el criterio?
                  /             \
                no               sí
                |                |
                +-> revisar      v
                          integración
                               |
                               v
                          monitoreo
                               |
                               +-> nuevos datos y revisión
```

| Etapa | Pregunta principal | Resultado esperado |
|---|---|---|
| Definición | ¿Qué decisión queremos mejorar? | Objetivo y contrato de salida |
| Datos | ¿Qué representa cada muestra? | Dataset documentado |
| Exploración | ¿Hay errores, sesgos o faltantes? | Informe de calidad |
| Baseline | ¿Qué logra una solución sencilla? | Punto de comparación |
| Entrenamiento | ¿Qué parámetros funcionan mejor? | Modelo versionado |
| Evaluación | ¿Generaliza y qué errores comete? | Métricas y análisis |
| Integración | ¿Cómo lo consume la aplicación? | Servicio o componente |
| Monitoreo | ¿Sigue funcionando con datos nuevos? | Alertas y trazas |

Un proyecto no empieza eligiendo una arquitectura. Primero pregunta:

1. ¿Qué decisión o proceso queremos mejorar?
2. ¿Qué salida concreta necesita el software?
3. ¿Qué datos existen y qué representan?
4. ¿Cómo mediremos un error y su costo?
5. ¿Qué ocurre cuando el modelo no sabe?

### Baseline

Un baseline es una solución de referencia deliberadamente simple.
Puede ser:

- predecir siempre la clase más frecuente;
- usar el promedio histórico;
- implementar una regla razonable;
- utilizar un modelo preentrenado sin adaptación.

Si un modelo complejo no supera el baseline de forma consistente,
todavía no justificamos su complejidad.

### Evaluar no es solamente calcular accuracy

Debemos conocer:

- qué tipo de error se produce;
- en qué subgrupos o condiciones falla;
- cuánto cuesta un falso positivo o un falso negativo;
- si la entrada real se parece a los datos de entrenamiento;
- si existe una forma segura de rechazar o derivar un caso.

### CPU, GPU y dispositivo

- La CPU es el procesador general y ejecutará todas las prácticas.
- Una GPU es un procesador que puede acelerar operaciones sobre
  tensores, es decir, cálculos repetidos sobre arreglos numéricos.
  No es un requisito para este módulo.
- Detectar una GPU no significa que todo el código la use
  automáticamente.

PyTorch suele nombrar al lugar de ejecución como **dispositivo**:

| Dispositivo | Significado |
|---|---|
| `cpu` | Procesador general |
| `cuda` | GPU NVIDIA accesible mediante la plataforma CUDA |
| `mps` | GPU de equipos Apple accesible mediante Metal Performance Shaders |

Estos nombres indican dónde puede ejecutarse un tensor; no describen
un algoritmo diferente.

### Semillas y reproducibilidad

Algunos algoritmos usan números pseudoaleatorios para crear datos,
inicializar parámetros o mezclar muestras. Una semilla fija ayuda a
repetir un experimento.

```text
misma versión + mismos datos + misma configuración + misma semilla
                          |
                          v
               resultado aproximadamente repetible
```

Una semilla no resuelve todas las diferencias entre hardware o
versiones, pero elimina una fuente común de variación.

````python
import platform
import sys

import numpy as np
import pandas as pd
import sklearn

FAST_MODE = True
SEED = 42
rng = np.random.default_rng(SEED)

diagnostico = {
    "Python": sys.version.split()[0],
    "Sistema": platform.platform(),
    "NumPy": np.__version__,
    "Pandas": pd.__version__,
    "Scikit-learn": sklearn.__version__,
    "FAST_MODE": FAST_MODE,
}

try:
    import torch
    diagnostico["PyTorch"] = torch.__version__
    diagnostico["Dispositivo"] = (
        "cuda" if torch.cuda.is_available()
        else "mps" if torch.backends.mps.is_available()
        else "cpu"
    )
except ImportError:
    diagnostico["PyTorch"] = "no instalado"
    diagnostico["Dispositivo"] = "cpu"

pd.Series(diagnostico, name="valor").to_frame()
````

### Cómo leer el diagnóstico

La tabla anterior no mide potencia ni calidad de los modelos. Solo
confirma el contexto de ejecución.

- **Python** identifica el intérprete que ejecuta el kernel.
- **Versiones** permiten reproducir o diagnosticar incompatibilidades.
- **Dispositivo** informa si PyTorch detecta CPU, CUDA o MPS.
- **FAST_MODE** indica que usaremos datasets y ciclos reducidos.

Cuando una importación funciona en la terminal pero falla en el
notebook, frecuentemente la terminal y el kernel están usando
entornos diferentes.

```text
terminal -> entorno A -> paquete instalado
notebook -> entorno B -> paquete ausente
```

Por eso siempre verificamos el intérprete desde el propio notebook.

## 8. Experimento: reglas frente a un modelo aprendido

Un equipo quiere anticipar alertas de sensores. Cada fila del
dataset representa una máquina observada:

| Columna | Interpretación | Rol |
|---|---|---|
| `temperatura` | Temperatura medida | Feature |
| `vibracion` | Nivel de vibración | Feature |
| `antiguedad` | Años de uso | Feature |
| `alerta` | Caso que requiere atención | Label |

El primer enfoque escribe dos reglas:

```text
si temperatura > 78 -> alerta
si vibración > 5     -> alerta
en otro caso         -> normal
```

El segundo utiliza un árbol de decisión. Un árbol aprende una
secuencia de preguntas a partir de ejemplos:

```text
            ¿vibración > valor?
               /          \
             no            sí
             |             |
    ¿temperatura...?    alerta
```

### Qué vamos a comparar

1. Generamos un historial sintético.
2. Separamos train y test.
3. El árbol observa solamente train.
4. Las reglas y el árbol reciben exactamente el mismo test.
5. Comparamos accuracy sobre esos casos reservados.

El ejemplo es pequeño y artificial. Sirve para observar el proceso,
no para diseñar mantenimiento predictivo real.

````python
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt

def reglas(temperatura, vibracion, antiguedad):
    return int(temperatura > 78 or vibracion > 5)

n = 220
datos = pd.DataFrame({
    "temperatura": rng.uniform(55, 95, n),
    "vibracion": rng.uniform(0.5, 7.0, n),
    "antiguedad": rng.integers(1, 13, n),
})
patron = (
    ((datos["temperatura"] > 76) & (datos["vibracion"] > 3))
    | ((datos["antiguedad"] > 8) & (datos["vibracion"] > 4))
)
datos["alerta"] = patron.astype(int)

X = datos[["temperatura", "vibracion", "antiguedad"]]
y = datos["alerta"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, stratify=y, random_state=SEED
)

pred_reglas = np.array([
    reglas(f.temperatura, f.vibracion, f.antiguedad)
    for f in X_test.itertuples(index=False)
])
arbol = DecisionTreeClassifier(max_depth=3, random_state=SEED)
arbol.fit(X_train, y_train)
pred_arbol = arbol.predict(X_test)

pd.DataFrame({
    "sistema": ["Reglas escritas", "Árbol aprendido"],
    "accuracy_test": [
        accuracy_score(y_test, pred_reglas),
        accuracy_score(y_test, pred_arbol),
    ],
}).round(3)
````

````python
plt.figure(figsize=(13, 5))
plot_tree(
    arbol,
    feature_names=X.columns,
    class_names=["normal", "alerta"],
    filled=True,
    rounded=True,
    fontsize=9,
)
plt.title("Preguntas aprendidas por el árbol")
plt.show()
````

### Cómo interpretar la comparación

**Accuracy** es la proporción de predicciones correctas:

```text
casos correctos / total de casos evaluados
```

Una accuracy mayor del árbol indica que sus preguntas se ajustaron
mejor al patrón del dataset de prueba. No indica que:

- haya descubierto una causa física;
- funcione con sensores reales;
- todos los tipos de error tengan el mismo costo;
- siga funcionando si cambia el proceso que genera los datos.

| Aspecto | Reglas | Modelo aprendido |
|---|---|---|
| Origen de la lógica | Personas | Datos y algoritmo |
| Facilidad de explicación | Alta si hay pocas reglas | Depende del modelo |
| Adaptación a patrones combinados | Manual | Puede aprender combinaciones |
| Dependencia de datos históricos | Baja | Alta |
| Riesgo principal | Casos no contemplados | Sesgo y mala generalización |

La conclusión correcta no es “ML siempre gana”. Reglas y modelos
pueden complementarse:

```text
validaciones obligatorias -> reglas
predicción incierta        -> modelo
umbral de seguridad        -> regla
revisión de casos dudosos  -> persona
```

## 9. Del experimento a un sistema de software

Para integrar un modelo necesitamos definir un contrato tan claro
como el de cualquier servicio:

| Pregunta | Ejemplo |
|---|---|
| ¿Qué recibe? | Tres números con unidades conocidas |
| ¿Qué devuelve? | Clase, score y versión del modelo |
| ¿Qué valida? | Rangos, faltantes y tipos |
| ¿Cuánto puede tardar? | Presupuesto de latencia |
| ¿Qué hace si no sabe? | Rechazar o derivar a revisión |
| ¿Qué registra? | Entrada permitida, salida, tiempo y versión |

En producción también importan:

- **latencia:** tiempo que tarda una inferencia;
- **throughput:** cantidad de solicitudes procesadas;
- **privacidad:** qué datos pueden almacenarse o enviarse;
- **versionado:** qué modelo y datos produjeron una salida;
- **observabilidad:** métricas, logs y trazas para diagnosticar;
- **drift:** cambio de los datos reales respecto del entrenamiento;
- **fallback:** comportamiento seguro cuando el sistema falla.

Un notebook demuestra una idea. Un producto necesita contratos,
pruebas, permisos, monitoreo y responsables.
