## 1. ¿Qué es un agente?

Un **agente** es un sistema que combina un modelo con la capacidad de **actuar** para lograr un objetivo.

La diferencia clave está en el verbo: un modelo **responde**, un agente **actúa**.

| Modelo solo | Agente |
|---|---|
| Recibe texto y devuelve texto | Recibe una consulta y decide qué hacer |
| No ejecuta acciones | Puede llamar herramientas |
| No recuerda conversaciones | Puede tener memoria |
| No tiene un objetivo propio | Persigue un objetivo definido |
| No sabe cuándo pedir ayuda | Tiene reglas y control |

> Idea central: el modelo es el "cerebro" que propone. El agente es el sistema que decide, actúa y controla.

---
## 2. Chatbot vs. agente

No todo lo que conversa es un agente. La diferencia está en si el sistema **puede hacer algo** más allá de hablar.

| Chatbot | Agente |
|---|---|
| Solo conversa | Conversa **y** actúa |
| Devuelve texto | Puede consultar datos, calcular, buscar |
| No cambia el mundo | Puede ejecutar acciones |
| Ejemplo: responder preguntas frecuentes | Ejemplo: consultar un pedido y dar su estado |

Un chatbot puede ser parte de un agente. Pero un agente agrega la capacidad de **hacer**.

En la Clase 2 vamos a construir nuestro primer agente conversacional. Hoy entendemos qué piezas lo componen.

---
## 3. Las partes de un agente

Todo agente combina varias piezas. Cada una cumple un rol:

| Parte | Qué hace | Ejemplo |
|---|---|---|
| **Modelo** | Interpreta y propone | Decide qué herramienta usar |
| **Herramientas** | Ejecutan acciones | Buscar, calcular, consultar una API |
| **Memoria** | Recuerda contexto | Historial de la conversación |
| **Reglas** | Limitan el comportamiento | No revelar datos personales |
| **Control** | Decide cuándo actuar o pedir ayuda | Derivar a una persona |

En las próximas clases vamos a agregar estas piezas una por una:
- **Clase 2:** el modelo y la función `responder()`.
- **Clase 3:** las herramientas.
- **Clase 5:** la memoria.
- **Clase 9:** las reglas y el control (riesgos e ISO 42001).

---
## 4. El ciclo del agente

Un agente trabaja en un ciclo continuo:

```text
entrada → interpretación → decisión → acción → resultado → control
```

1. **Entrada:** llega una consulta.
2. **Interpretación:** el modelo entiende qué pide.
3. **Decisión:** decide qué hacer (responder, usar una herramienta, pedir ayuda).
4. **Acción:** ejecuta la herramienta o genera la respuesta.
5. **Resultado:** obtiene el dato o el texto.
6. **Control:** verifica si el resultado es suficiente o si debe intervenir una persona.

En la celda siguiente vemos este ciclo en acción con un agente muy simple.

---
## 5. Mini-demo: un agente que decide

Vamos a construir un agente muy simple que decide qué hacer según la consulta. No usa un modelo real todavía: usa reglas. Pero muestra el ciclo completo.

El agente tiene tres posibles acciones:
- responder directamente,
- usar una herramienta (consultar un pedido),
- pedir ayuda a una persona.

En la celda siguiente lo vemos en código, paso a paso.

````python
# Mini-demo: un agente simple que decide qué hacer.

# Herramienta: consultar el estado de un pedido.
PEDIDOS = {"A100": "en preparación", "A101": "en camino", "A102": "entregado"}

def consultar_pedido(codigo):
    return PEDIDOS.get(codigo.upper(), "pedido no encontrado")

# El agente: interpreta, decide, actúa y controla.
def agente(consulta):
    texto = consulta.lower()

    # 1. INTERPRETACIÓN: ¿qué pide?
    if "pedido" in texto:
        # 2. DECISIÓN: usar la herramienta de pedidos.
        # 3. ACCIÓN: ejecutar la herramienta.
        resultado = consultar_pedido("A101")
        # 4. RESULTADO + CONTROL: devolver la respuesta.
        return f"Tu pedido está {resultado}."

    if "precio" in texto:
        return "Puedo ayudarte con información comercial."

    # Si no sabe qué hacer, pide ayuda (control).
    return "Necesito más información. ¿Podés detallar tu consulta?"

# Probamos el ciclo completo.
consultas = ["¿Dónde está mi pedido?", "¿Cuál es el precio?", "Hola"]
for consulta in consultas:
    print("PERSONA:", consulta)
    print("AGENTE:", agente(consulta), "\n")
````

### ¿Dónde está cada parte del agente?

En el código anterior podemos identificar cada pieza:

| Parte | Dónde está en el código |
|---|---|
| **Modelo** | La lógica que interpreta la consulta (los `if`) |
| **Herramienta** | La función `consultar_pedido()` |
| **Memoria** | Todavía no hay (no recuerda) |
| **Reglas** | Las condiciones que limitan qué responde |
| **Control** | El `return` final que pide más información |

> Nota: en este ejemplo el "modelo" son reglas escritas a mano. En la Clase 2 lo reemplazaremos por una función `responder()` que puede usar un modelo real.

## 6. Incorporar un motor LLM real

Hasta acá el "modelo" del agente eran reglas escritas a mano (los `if`). Ahora vamos a reemplazarlo por un **motor LLM real**: un modelo de lenguaje que interpreta la consulta y decide qué hacer.

Usaremos un **modelo local** con llama.cpp. No hace falta internet ni API key: el modelo se descarga una vez y corre en tu máquina.

| Qué necesitás | Detalle |
|---|---|
| `llama-cpp-python` | Librería para correr modelos GGUF |
| `huggingface-hub` | Para descargar el modelo |
| Modelo GGUF | Granite-4.0-350M (pequeño, corre en CPU) |

> Este es el mismo patrón del wrapper `llamar_llm()` que usamos en el módulo 2. La idea es tener una única función que envía un prompt y devuelve texto.

````python
# ─── Instalación de dependencias (solo la primera vez) ────────────────────────
# Descomentá esta línea si no tenés las librerías instaladas:
# !pip install llama-cpp-python huggingface-hub
print("Listo para configurar.")
````

````python
# ─── Inicializar el modelo local ─────────────────────────────────────────────

import time
from huggingface_hub import hf_hub_download
from llama_cpp import Llama

# Qwen 3.5 + Claude Opus 4.6 destilado — 1.2 GB de peso
#REPO_ID = "Jackrong/Qwen3.5-2B-Claude-4.6-Opus-Reasoning-Distilled-GGUF"
#FILENAME = "Qwen3.5-2B.Q4_K_M.gguf"


REPO_ID = "unsloth/LFM2.5-1.2B-Instruct-GGUF"
FILENAME = "LFM2.5-1.2B-Instruct-Q8_0.gguf"

print("Descargando modelo local (puede tardar la primera vez)...")

ruta_modelo = hf_hub_download(repo_id=REPO_ID, filename=FILENAME)

_llm_local = Llama(model_path=ruta_modelo, n_ctx=2048, n_gpu_layers=0, verbose=False)

print("\n Modelo local listo.")
````

````python
def llamar_llm(
    prompt,
    system_prompt="Sos un asistente de atención. Respondé con claridad y brevedad.",
    temperature=0.7,
    max_tokens=200
):
    """Envía un prompt al modelo local y devuelve la respuesta como string."""

    respuesta = _llm_local.create_chat_completion(
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user",   "content": prompt}
        ],
        temperature=temperature,
        max_tokens=max_tokens
    )
    return respuesta["choices"][0]["message"]["content"].strip()


# Prueba rápida de conexión
print(llamar_llm("Decí 'Hola, estoy funcionando'", max_tokens=20))
````

### El agente con motor LLM real

Ahora el "modelo" del agente ya no son reglas escritas a mano. Es un LLM real que interpreta la consulta y decide qué hacer.

En la celda siguiente vemos cómo el agente usa `llamar_llm()` para responder. El modelo interpreta la consulta y genera una respuesta natural, en vez de depender de `if` fijos.

````python
# El agente ahora usa un motor LLM real para interpretar y responder.

def agente_llm(consulta):
    # El modelo interpreta la consulta y genera la respuesta.
    return llamar_llm(consulta, max_tokens=120)

# Probamos el agente con motor LLM.
consultas = ["¿Dónde está mi pedido?", "¿Cuál es el precio?", "Hola"]
for consulta in consultas:
    print("PERSONA:", consulta)
    print("AGENTE:", agente_llm(consulta), "\n")
````
